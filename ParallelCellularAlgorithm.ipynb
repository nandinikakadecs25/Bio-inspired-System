```python

import numpy as np
import matplotlib.pyplot as plt

# Parallel Cellular Algorithm
# Smart Traffic Management System

# Initial traffic density of 5 junctions
traffic = np.array([40, 50, 60, 30, 70])

# Adjustment factor
alpha = 0.25

# Calculate neighbouring average (circular neighbourhood)
neighbor_avg = (
    np.roll(traffic, 1) + np.roll(traffic, -1)
) / 2

# Calculate change in traffic density
delta_x = alpha * (neighbor_avg - traffic)

# Update traffic density
new_traffic = traffic + delta_x

# Display results
print("PARALLEL CELLULAR ALGORITHM")
print("--------------------------------")

for i in range(5):
    print("Junction", i + 1)
    print("Initial Density:", traffic[i])
    print("Neighbour Average:", neighbor_avg[i])
    print("Change:", delta_x[i])
    print("New Density:", new_traffic[i])
    print("--------------------------------")

# Create graph
junctions = np.arange(1, 6)

plt.figure(figsize=(10, 5))

plt.plot(junctions, traffic, marker='o',
         label='Initial Traffic Density')

plt.plot(junctions, new_traffic, marker='s',
         label='Updated Traffic Density')

plt.xlabel("Traffic Junctions")
plt.ylabel("Number of Vehicles")
plt.title("Parallel Cellular Algorithm - Traffic Management")

plt.xticks(junctions)
plt.grid(True)
plt.legend()

plt.show()

# Bar graph
plt.figure(figsize=(10, 5))

width = 0.35

plt.bar(junctions - width/2, traffic, width,
        label='Before Update')

plt.bar(junctions + width/2, new_traffic, width,
        label='After Update')

plt.xlabel("Traffic Junctions")
plt.ylabel("Number of Vehicles")
plt.title("Traffic Density Comparison")

plt.xticks(junctions)
plt.legend()
plt.grid(axis='y')

plt.show()
```
